### Description of Attributes in the Heart Disease Dataset
[heart dataset](https://www.kaggle.com/datasets/arezaei81/heartcsv/data)

Here is a breakdown of the 14 attributes (features and target) present in your `heart.csv` dataset:

1. **`age`**: Age of the patient in years.
2. **`sex`**: Gender of the patient:
   * `1` = Male
   * `0` = Female
3. **`cp`**: Chest Pain type:
   * `0` = Typical Angina (chest pain related to decreased blood flow to the heart)
   * `1` = Atypical Angina (chest pain not related to heart issues)
   * `2` = Non-anginal Pain (typically esophageal spasms or non-cardiac)
   * `3` = Asymptomatic (silent, but clinically significant)
4. **`trestbps`**: Resting blood pressure in mm Hg on admission to the hospital.
5. **`chol`**: Serum cholesterol level in mg/dl.
6. **`fbs`**: Fasting blood sugar levels > 120 mg/dl:
   * `1` = True
   * `0` = False
7. **`restecg`**: Resting electrocardiographic results:
   * `0` = Normal
   * `1` = Having ST-T wave abnormality (T wave inversions and/or ST elevation or depression of > 0.05 mV)
   * `2` = Showing probable or definite left ventricular hypertrophy by Estes' criteria
8. **`thalach`**: Maximum heart rate achieved during exercise.
9. **`exang`**: Exercise-induced angina (chest pain triggered by physical exertion):
   * `1` = Yes
   * `0` = No
10. **`oldpeak`**: ST depression induced by exercise relative to rest (measures electrical abnormality/strain on the heart during physical effort).
11. **`slope`**: The slope of the peak exercise ST segment:
    * `0` = Upsloping (generally healthy heart response)
    * `1` = Flat (possible heart disease marker)
    * `2` = Downsloping (stronger indicator of heart disease)
12. **`ca`**: Number of major vessels (0–4) colored by fluoroscopy (indicates blood vessel blockage; higher numbers mean more blocked vessels).
13. **`thal`**: Thalassemia (a blood disorder):
    * `1` = Fixed defect (no blood flow in some part of the heart)
    * `2` = Normal blood flow
    * `3` = Reversible defect (blood flow is observed but not fully normal)
14. **`target`**: Heart Disease status (Target Variable):
    * `1` = Defective Heart (Heart Disease detected)
    * `0` = Healthy Heart (No Heart Disease)

### The Mathematical Foundation of Logistic Regression

Unlike Linear Regression which predicts continuous values, **Logistic Regression** is used for classification (predicting probabilities between 0 and 1). Here is how the mathematics works:

#### 1. The Linear Equation
First, the algorithm takes your input features ($X_1, X_2, ..., X_n$, such as `age`, `chol`, `cp`) and calculates a weighted sum, similar to linear regression:

$$z = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + ... + \beta_n X_n$$

Where:
* $\beta_0$ is the bias (intercept).
* $\beta_1, \beta_2, ..., \beta_n$ are the coefficients (weights) the model learns during training.
* $z$ can range from negative infinity ($-\infty$) to positive infinity ($+\infty$).

#### 2. The Sigmoid Function (Activation)
To convert $z$ into a probability value between $0$ and $1$, we pass it through the **Sigmoid (or Logistic) function**:

$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

Where:
* $e$ is Euler's number (approx. 2.71828).
* If $z$ is a large positive number, $e^{-z} \to 0$, so $\sigma(z) \to 1$.
* If $z$ is a large negative number, $e^{-z} \to \infty$, so $\sigma(z) \to 0$.
* If $z = 0$, then $\sigma(z) = 0.5$.

This maps any input $z$ to a probability $P(Y=1|X) = \sigma(z)$.

#### 3. The Decision Boundary
By default, a threshold of $0.5$ is set to classify predictions:
* If $P(Y=1|X) \ge 0.5$ (which happens when $z \ge 0$), predict class **1** (Defective Heart).
* If $P(Y=1|X) < 0.5$ (which happens when $z < 0$), predict class **0** (Healthy Heart).

#### 4. The Cost Function: Binary Cross-Entropy (Log Loss)
To learn the best weights ($\beta$), the model minimizes the difference between its predicted probabilities ($p_i$) and the actual labels ($y_i \in \{0, 1\}$). We use the **Log Loss** function:

$$J(\beta) = -\frac{1}{m} \sum_{i=1}^{m} \left[ y_i \log(p_i) + (1 - y_i) \log(1 - p_i) \right]$$

* **When actual $y_i = 1$:** The right part of the equation becomes 0. If the predicted probability $p_i \to 1$, the loss $-\log(p_i) \to 0$. If $p_i \to 0$, the loss $-\log(p_i) \to \infty$.
* **When actual $y_i = 0$:** The left part becomes 0. If predicted $p_i \to 0$, the loss $-\log(1 - p_i) \to 0$. If $p_i \to 1$, the loss $-\log(1 - p_i) \to \infty$.

#### 5. Optimization: Gradient Descent
The algorithm iteratively updates the weights $\beta$ using **Gradient Descent** to find the minimum of the cost function $J(\beta)$:

$$\beta_j := \beta_j - \alpha \frac{\partial J(\beta)}{\partial \beta_j}$$

Where $\alpha$ is the learning rate, and the gradient is calculated as:

$$\frac{\partial J(\beta)}{\partial \beta_j} = \frac{1}{m} \sum_{i=1}^{m} (p_i - y_i) X_{ij}$$

### The Mathematical Foundation of Random Forest

**Random Forest** is an ensemble learning method primarily used for classification and regression. Instead of relying on a single decision tree, it constructs a "forest" of multiple decision trees and aggregates their predictions.

#### 1. Ensemble Method: Bagging (Bootstrap Aggregating)
To train individual decision trees, Random Forest uses a statistical technique called **Bagging** to ensure diversity among the models:
* **Bootstrapping:** If your training set has $N$ samples, Random Forest creates multiple subsets of size $N$ by sampling *with replacement*. This means some data points will be repeated in a subset, while others (roughly $36.8\%$, known as **Out-of-Bag** samples) are left out.
* **Aggregating:** For classification, the final prediction is decided by a **majority vote** from all the individual trees:
  $$\hat{Y} = \text{mode}\{T_1(X), T_2(X), ..., T_B(X)\}$$
  where $B$ is the number of trees (`n_estimators=100`) and $T_i(X)$ is the class predicted by the $i$-th tree.

#### 2. Feature Randomness (The "Random" in Random Forest)
In a standard decision tree, the algorithm searches through *all* available features to find the best split. In Random Forest, each tree is allowed to search only through a **random subset of features** at each split:
* For classification, if you have $p$ total features, the default subset size is typically $\lfloor \sqrt{p} \rfloor$.
* For this dataset with $13$ features, each node split only evaluates about $\sqrt{13} \approx 3$ randomly selected features. This mathematically decorrelates the trees so that a single highly dominant feature does not make all trees look identical.

#### 3. Mathematical Splitting Criteria (How Trees Learn)
At each node of an individual tree, the algorithm determines the optimal feature and split threshold by maximizing **Information Gain** or minimizing **Gini Impurity**.

##### **Gini Impurity ($I_G$):**
Measures the probability of a randomly chosen element from the set being incorrectly labeled if it was randomly labeled according to the distribution of labels in the subset.

$$I_G(p) = 1 - \sum_{i=1}^{C} p_i^2$$

Where:
* $C$ is the number of classes ($C = 2$ for target `0` and `1`).
* $p_i$ is the probability/proportion of samples belonging to class $i$ in that node.
* A pure node (containing only healthy or only defective hearts) has $I_G = 0$.

##### **Information Gain ($IG$):**
When splitting a parent node ($D_p$) into left ($D_l$) and right ($D_r$) children nodes, the algorithm selects the split that yields the highest reduction in impurity:

$$IG(D_p, f) = I_G(D_p) - \left( \frac{N_l}{N_p} I_G(D_l) + \frac{N_r}{N_p} I_G(D_r) \right)$$

Where:
* $f$ is the feature being split on.
* $N_p, N_l, N_r$ are the number of samples in the parent, left child, and right child nodes, respectively.

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
```

```python
# Feature Scaling
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
# Train Test Split

X_train_lstm, X_test_lstm, Y_train_lstm, Y_test_lstm = train_test_split(
    X_scaled,
    Y,
    test_size=0.2,
    random_state=2
)

print("Before Reshape:")
print(X_train_lstm.shape)
print(X_test_lstm.shape)
```

```python
# Reshape Data For LSTM


X_train_lstm = X_train_lstm.reshape(
    X_train_lstm.shape[0],
    X_train_lstm.shape[1],
    1
)

X_test_lstm = X_test_lstm.reshape(
    X_test_lstm.shape[0],
    X_test_lstm.shape[1],
    1
)

print("After Reshape:")
print(X_train_lstm.shape)
print(X_test_lstm.shape)
```

### The Mathematical Foundation of LSTMs (Long Short-Term Memory)

Standard Recurrent Neural Networks (RNNs) struggle with **vanishing and exploding gradients** when processing longer sequences because backpropagation gradients must pass through continuous matrix multiplications over time. LSTMs solve this bottleneck by introducing an additive memory bridge called the **cell state**, regulated by three specialized gates.

---

### 1. The Core Innovation: Cell State vs. Hidden State
At any given time step $t$, the LSTM maintains two vectors:
* **Cell State ($C_t$):** The long-term memory corridor. Information flows down the cell state with only minor linear interactions, making it easy for gradients to flow backward through time without vanishing.
* **Hidden State ($h_t$):** The short-term memory vector. It acts as the immediate output of the cell at step $t$ and is used for predictions.

---

### 2. Step-by-Step Gate Operations
An LSTM cell uses three gates to control the flow of information. Each gate utilizes a Sigmoid activation function ($\sigma$) to output values between $0$ (completely block) and $1$ (completely allow).

#### A. The Forget Gate ($f_t$)
Decides what information from the previous cell state $C_{t-1}$ should be discarded. It looks at the current feature input $x_t$ and the previous short-term state $h_{t-1}$:

$$f_t = \sigma\left(W_f \cdot [h_{t-1}, x_t] + b_f\right)$$

Where:
* $W_f$ and $b_f$ are the weights and bias learned during training.
* $\sigma(z) = \frac{1}{1 + e^{-z}}$ limits the output scale between $0$ and $1$.

#### B. The Input Gate ($i_t$) and Candidate State ($\tilde{C}_t$)
Decides what new information should be added to the cell state. This is split into two parts:
1. **The Input Gate ($i_t$):** Determines which indices of the state vector to update.
   $$i_t = \sigma\left(W_i \cdot [h_{t-1}, x_t] + b_i\right)$$
2. **The Candidate State ($\tilde{C}_t$):** Generates new candidate values to store in the cell state using a Hyperbolic Tangent (tanh) activation function to scale values between $-1$ and $1$.
   $$\tilde{C}_t = \tanh\left(W_c \cdot [h_{t-1}, x_t] + b_c\right)$$

#### C. Updating the Cell State ($C_t$)
We combine the decisions of the forget gate and input gate to update the old memory state $C_{t-1}$ into the new memory state $C_t$:

$$C_t = f_t \odot C_{t-1} + i_t \odot \tilde{C}_t$$

Where $\odot$ represents element-wise (Hadamard) multiplication. This linear addition is key to protecting gradients from vanishing.

#### D. The Output Gate ($o_t$) and Final Hidden State ($h_t$)
Decides what the next hidden state (short-term output) should be:
1. **The Output Gate ($o_t$):** Determines which parts of the cell state we are going to output.
   $$o_t = \sigma\left(W_o \cdot [h_{t-1}, x_t] + b_o\right)$$
2. **The Hidden State ($h_t$):** We pass the newly updated cell state through $\tanh$ (to normalize values back between $-1$ and $1$) and multiply it by the output gate:
   $$h_t = o_t \odot \tanh(C_t)$$
